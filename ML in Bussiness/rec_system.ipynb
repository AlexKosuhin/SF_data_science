```python
articles_df = pd.read_csv(r'C:\Users\user\Desktop\Skilfactory\data1\shared_articles.csv')
articles_df = articles_df[articles_df['eventType'] == 'CONTENT SHARED']
articles_df.shape
```

```python
interactions_df = pd.read_csv(r'C:\Users\user\Desktop\Skilfactory\data1\users_interactions.csv')
```

```python
interactions_df.personId = interactions_df.personId.astype(str)
interactions_df.contentId = interactions_df.contentId.astype(str)
articles_df.contentId = articles_df.contentId.astype(str)
```

```python
interactions_df.isnull().sum()
```

```python
event_type = {
   'VIEW': 1.0,
   'LIKE': 2.0, 
   'BOOKMARK': 2.5, 
   'FOLLOW': 3.0,
   'COMMENT CREATED': 4.0,  
}
```

```python
interactions_df['eventType'] = interactions_df['eventType'].map(event_type)
interactions_df['eventType'].mean(skipna=True)
```

```python
interactions_df.info()
```

```python
users_interactions_count_df = (
    interactions_df
    .groupby(['personId', 'contentId'])
    .first()
    .reset_index()
    .groupby('personId').size())

users_with_enough_interactions_df = \
    users_interactions_count_df[users_interactions_count_df >= 5].reset_index()[['personId']]

print(len(users_with_enough_interactions_df))
```

```python
interactions_from_selected_users_df = interactions_df.loc[np.in1d(interactions_df.personId,
            users_with_enough_interactions_df)]
print(interactions_from_selected_users_df.shape)
```
