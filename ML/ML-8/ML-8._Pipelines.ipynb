# ML-8. Pipelines

```python
#загрузим основные библиотеки
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
 
from sklearn.datasets import fetch_california_housing
from sklearn import metrics
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split, cross_val_score
```

Попробуем предсказать цену на недвижимость в Калифорнии

## Часть 1. Простейшие пайплайны


Загрузим данные

```python
data = fetch_california_housing()
```

```python
df = pd.DataFrame(data['data'], columns=data['feature_names'])
df.loc[:,'target'] = data['target']
df.describe()
```

```python
def rmse(y_hat, y):
    return metrics.mean_squared_error(y_hat, y, squared = False)
```

```python
X = df.drop('target', axis=1)
Y = df['target']
X_train, X_test, y_train, y_test = train_test_split(X, Y, random_state=42)
```

```python
print(f'Размер обучающей выборки {X_train.shape}')
print(f'Размер тестовой выборки {X_test.shape}')
```

```python
pipeline = Pipeline([('scaler', StandardScaler()), ('rf', RandomForestRegressor())])
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
print(f'Качество по метрике R2: { round(r2_score(y_test, y_pred),4)}')
rmse_value = np.sqrt(mean_squared_error(y_test, y_pred))
print(f'Качество по RMSE: {round(rmse_value, 4)}')
```

```python
pipeline.get_params()
```

```python
print(pipeline[1].n_estimators)
print(pipeline['rf'].n_estimators)
```

```python
pipeline.set_params(rf__n_estimators=200)
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {'scaler__with_mean':[True,False],
              'rf__n_estimators':[100, 200, 500]}
grid_search = GridSearchCV(pipeline, param_grid=param_grid, verbose = True)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import make_column_transformer
ct = make_column_transformer(
     (StandardScaler(), ['Price']),
    (OneHotEncoder(), ['Country']))
print(ct)
```

```python
pipeline = Pipeline([('ct', ct), ('rf', RandomForestRegressor())])
```

```python
X = df_wine[['Country', 'Price']]
y = df_wine['Rating']
```

```python
pd.DataFrame(pipeline['ct'].transform(X).toarray(), columns = ['Price']+ pipeline['ct'].transformers_[1][1].get_feature_names_out().tolist())
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.compose import make_column_transformer
ct = make_column_transformer(
    (OrdinalEncoder(), ['Region']),
    (StandardScaler(), ['Price']),
    (OneHotEncoder(), ['Country'])
)
print(ct)
```

```python
pipeline_new = Pipeline([('ct', ct), ('rf', RandomForestRegressor(random_state=42))])
X = df_wine[['Country', 'Price', 'Region']]
y = df_wine['Rating']
pipeline_new.fit(X, y)
```

```python
df_wine_test = pd.read_csv(r'C:\Users\user\Desktop\Skilfactory\data1\Red_test.csv')
```

```python
X_test = df_wine_test[['Country', 'Price', 'Region']]
y_test = df_wine_test['Rating']
```

```python
y_predict_test = pipeline_new.predict(X_test)
rmse_value = np.sqrt(mean_squared_error(y_test, y_predict_test))
print(f'Качество по RMSE: {round(rmse_value, 4)}')
```

```python
joblib.dump(pipeline_new, 'pipeline_new.pkl')
```

```python
pipeline_new.set_params(rf__n_estimators=200)
pipeline_new.fit(X, y)
```

```python
y_predict_test = pipeline_new.predict(X_test)
rmse_value = np.sqrt(mean_squared_error(y_test, y_predict_test))
print(f'Качество по RMSE: {round(rmse_value, 4)}')
```

```python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import RidgeCV
from sklearn.tree import DecisionTreeRegressor
estimators = [
    ('ridge', RidgeCV()),
    ('dt', DecisionTreeRegressor(random_state=42))
]
reg = StackingRegressor(
    estimators=estimators,
    final_estimator=RandomForestRegressor(n_estimators=10, random_state=42)
)
pipeline_3 = Pipeline([('ct', ct), ('rf', reg)])
```

```python
pipeline_3.fit(X, y)
```

```python
y_predict_test = pipeline_3.predict(X_test)
rmse_value = np.sqrt(mean_squared_error(y_test, y_predict_test))
print(f'Качество по RMSE: {round(rmse_value, 4)}')
```
