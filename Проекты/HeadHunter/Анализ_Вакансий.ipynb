Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
def work_experience(x):
    if pd.isna(x) or x == 'Не указано':
        return np.nan
    else:
        list_x = x.split('  ')[0].split()
        if len(list_x) == 6:
            year = int(list_x[2])
            month = int(list_x[4])
            return year * 12 + month
        elif list_x[-1] == 'года' or list_x[-1] == 'год' or list_x[-1] == 'лет':
            year = int(list_x[-2])
            return year * 12
        elif list_x[-1] == 'месяцев' or list_x[-1] == 'месяц' or list_x[-1] == 'месяца':
            month = int(list_x[-2])
            return month


df['Опыт работы (месяц)'] = df['Опыт работы'].apply(work_experience)
print(f'медианный опыт работы: {np.median(df['Опыт работы (месяц)'].dropna())} месяцев')
df.drop('Опыт работы', axis=1, inplace=True)
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
def town(x):
    list_x = x.split()
    city = list_x[0]
    million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 
                      'Казань', 'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 
                      'Уфа', 'Красноярск', 'Пермь', 'Воронеж', 'Волгоград' ]
    if city == 'Москва':
        return 'Москва'
    elif city == 'Санкт-Петербург':
            return 'Санкт-Петербург'
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'


df['Город'] = df['Город, переезд, командировки'].apply(town)
df['Город'].value_counts()
```

```python
def removal(x):
    list_x = x.split(' , ')[1:3]
    if ' м. ' in list_x[0]:
        travel = list_x[1] # Определяем местоположение необходимой нам информации
    else:
        travel = list_x[0] # Определяем местоположение необходимой нам информации
        
    if '(' in travel: # Избавляемся от городов в скобке
        ind = travel.index('(')
        travel_clean = travel[:ind].strip()
    else:
        travel_clean = travel
    
    if travel_clean in ['готов к переезду',
       'готова к переезду', 'хочу переехать']:
        return True
    else:
        return False
    
    
df['Готовность к переезду'] = df['Город, переезд, командировки'].apply(removal)
df['Готовность к переезду'].unique()
```

```python
def business_trip(x):
    trip = x.split(' , ')[-1]
    
    if '(' in trip: # Избавляемся от городов в скобке
        ind = trip.index('(')
        trip_clean = trip[:ind].strip()
    else:
        trip_clean = trip
    
    if trip_clean in ['готова к командировкам', 'готов к редким командировкам',
       'готов к командировкам', 'готова к редким командировкам', 'готова к переезду' , 'готов к переезду']:
        return True
    else:
        return False
    
    
df['Готовность к командировкам'] = df['Город, переезд, командировки'].apply(business_trip)
df.drop(['Город, переезд, командировки'], axis=1, inplace=True)
df['Готовность к командировкам'].unique()
```

```python
SPB = df['Город'].value_counts().get('Санкт-Петербург', 0)
print(f'Процент соискателей, проживающих в Санкт-Петербурге: {round((SPB / df.shape[0]), 2) * 100} %')

count_travel_and_business_trip = df[df['Готовность к командировкам'] & df['Готовность к переезду']].shape[0]
print(f'Процент соискателей, готовых одновременно и к переездам, и к командировкам: {round((count_travel_and_business_trip / df.shape[0]), 2) * 100} %')
```

Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
df_forex = pd.read_csv(r'C:\Users\user\Desktop\Skilfactory\data\ExchangeRates.csv')
df_forex.drop(['per', 'time', 'vol'], axis=1, inplace=True)

df['Обновление резюме'] = pd.to_datetime(df['Обновление резюме'], dayfirst=True)
df['Обновление резюме'] = df['Обновление резюме'].dt.date
df['Обновление резюме'] = pd.to_datetime(df['Обновление резюме'], dayfirst=True)
df_forex['date'] = (pd.to_datetime(df_forex['date'], dayfirst=True))

df[['Сумма ЗП', 'Валюта']] = df['ЗП'].str.split(expand=True)
Dict = {
    "грн.": "UAH",
    "USD": "USD",
    "EUR": "EUR",
    "бел.руб.": "BYN",
    "KGS": "KGS",
    "сум": "UZS",
    "AZN": "AZN",
    "KZT": "KZT",
    "руб.": "RUB"
}
df['Валюта'] = df['Валюта'].map(Dict)
df_merge = df.merge(df_forex, how='left', left_on=['Обновление резюме', 'Валюта'], right_on=['date', 'currency'])
df_merge.loc[df_merge['Валюта'] == 'RUB', 'close'] = 1

df_merge['ЗП (руб.)'] = df_merge.apply(lambda row: int(row['Сумма ЗП']) if row['Валюта'] == 'RUB'
                                       else int(row['Сумма ЗП']) / row['proportion'] * row['close'], axis=1)

print(f'Медианная зарплата: {np.median(df_merge["ЗП (руб.)"])} руб')
df_merge.drop(['Сумма ЗП', 'Валюта', 'currency', 'date', 'close', 'proportion', 'ЗП'], axis=1, inplace=True)
```

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
log_age = np.log(df_clean['Возраст'] + 1)
std_age = np.std(log_age)
mean_age = log_age.mean()
sigma_3 = mean_age - 3 * std_age
sigma_4 = mean_age + 4 * std_age
histplot = sns.histplot(log_age,
             bins=100)
histplot.set_title('Распределение возраста с логарифмированием')
histplot.set_xlabel('Возраст (логарифмированный)')
histplot.set_ylabel('Частота')
histplot.axvline(mean_age, color='red', lw=3)
histplot.axvline(sigma_3, color='red', lw=2)
histplot.axvline(sigma_4, color='red', lw=2);
```

```python
def outliers_z_score(data, feature, log_scale=False):
  if log_scale:
    x = np.log(data[feature] + 1)
  else:
    x = data[feature]

  mu = np.mean(x)
  sigma = np.std(x)
  lower_bound = mu - 3 * sigma
  upper_bound = mu + 4 * sigma
  outliers = data[(x < lower_bound) | (x > upper_bound)]
  cleaned = data[(x >= lower_bound) & (x <= upper_bound)]
  return outliers, cleaned

outliers, cleaned = outliers_z_score(df_clean, 'Возраст', log_scale=True)
print(f'Число выбросов по методу z-отклонения: {outliers.shape[0]}')
print(f'Результирующее число записей: {cleaned.shape[0]}')
```
