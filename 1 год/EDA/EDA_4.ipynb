```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns

# Настройки стиля
plt.style.use('default')
sns.set_palette("husl")

# Создаем данные
np.random.seed(42)
data = np.random.normal(0, 1, 1000)  # Нормальное распределение

# Параметры
alpha = 0.05
test_value = 0.0  # Проверяем отличие от 0

# Вычисляем p-value для разных тестов
p_value_two_sided = stats.ttest_1samp(data, test_value).pvalue
p_value_right = stats.ttest_1samp(data, test_value, alternative='greater').pvalue
p_value_left = stats.ttest_1samp(data, test_value, alternative='less').pvalue

print(f"Двухсторонний p-value: {p_value_two_sided:.4f}")
print(f"Правосторонний p-value: {p_value_right:.4f}") 
print(f"Левосторонний p-value: {p_value_left:.4f}")

# Создаем график
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Функция для визуализации одного теста
def plot_test(ax, p_value, test_type, alpha=0.05):
    # Рисуем распределение
    x = np.linspace(-4, 4, 1000)
    y = stats.norm.pdf(x)
    ax.plot(x, y, 'b-', linewidth=2)
    
    # Критические области
    if test_type == 'two-sided':
        left_critical = stats.norm.ppf(alpha/2)
        right_critical = stats.norm.ppf(1 - alpha/2)
        ax.fill_between(x, 0, y, where=(x <= left_critical), color='red', alpha=0.3, label='Критическая область')
        ax.fill_between(x, 0, y, where=(x >= right_critical), color='red', alpha=0.3)
    elif test_type == 'right':
        critical = stats.norm.ppf(1 - alpha)
        ax.fill_between(x, 0, y, where=(x >= critical), color='red', alpha=0.3, label='Критическая область')
    elif test_type == 'left':
        critical = stats.norm.ppf(alpha)
        ax.fill_between(x, 0, y, where=(x <= critical), color='red', alpha=0.3, label='Критическая область')
    
    # Вывод решения
    decision = "Отвергаем H₀" if p_value < alpha else "Не отвергаем H₀"
    color = 'green' if p_value >= alpha else 'red'
    
    ax.set_title(f'{test_type.capitalize()} тест\np-value = {p_value:.4f}, α = {alpha}\n{decision}', 
                 color=color, fontweight='bold')
    ax.set_xlabel('Z-статистика')
    ax.set_ylabel('Плотность вероятности')
    ax.legend()
    ax.grid(True, alpha=0.3)

# Рисуем все три теста
plot_test(axes[0], p_value_two_sided, 'two-sided')
plot_test(axes[1], p_value_right, 'right')  
plot_test(axes[2], p_value_left, 'left')

plt.tight_layout()
plt.show()

# Практический пример
print("\n" + "="*50)
print("ПРАКТИЧЕСКИЙ ПРИМЕР:")
print("="*50)

# Допустим, тестируем новое лекарство
group_a = np.random.normal(5.2, 1.5, 100)  # Контрольная группа
group_b = np.random.normal(5.8, 1.5, 100)  # Экспериментальная группа

print("Группа A (контроль): mean =", round(np.mean(group_a), 2))
print("Группа B (эксперимент): mean =", round(np.mean(group_b), 2))

# Разные типы тестов
t_stat, p_two = stats.ttest_ind(group_a, group_b)
_, p_right = stats.ttest_ind(group_a, group_b, alternative='less')  # B > A?
_, p_left = stats.ttest_ind(group_a, group_b, alternative='greater') # B < A?

print(f"\nДвухсторонний: p = {p_two:.4f} - есть ли разница?")
print(f"Правосторонний: p = {p_right:.4f} - B лучше A?")  
print(f"Левосторонний: p = {p_left:.4f} - B хуже A?")

# Решения
print(f"\nПри α = {alpha}:")
print(f"Двухсторонний: {'Отвергаем H₀ - есть разница' if p_two < alpha else 'Не отвергаем H₀ - разницы нет'}")
print(f"Правосторонний: {'Отвергаем H₀ - B лучше A' if p_right < alpha else 'Не отвергаем H₀ - B не лучше A'}")
print(f"Левосторонний: {'Отвергаем H₀ - B хуже A' if p_left < alpha else 'Не отвергаем H₀ - B не хуже A'}")
```

```python
# python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

def plot_hypothesis_test(test='z', tails='two', alpha=0.05, stat=2.1, df=None):
    """
    Рисует распределение H0, критические области, p-value и решение.
    Параметры:
      test: 'z' или 't'
      tails: 'two' (двусторонний), 'left' (левосторонний), 'right' (правосторонний)
      alpha: уровень значимости
      stat: наблюдаемое значение статистики теста (z или t)
      df: степени свободы для t-теста (обязательно, если test='t')
    """
    # Выбор распределения под H0
    if test == 'z':
        dist = stats.norm
        # Граница по оси X для наглядности
        x_min, x_max = -4, 4
    elif test == 't':
        if df is None:
            raise ValueError("Для t-теста укажите df (степени свободы).")
        dist = stats.t(df)
        # Адаптивные границы
        x_min = dist.ppf(1e-3)
        x_max = dist.ppf(1 - 1e-3)
    else:
        raise ValueError("test должен быть 'z' или 't'.")

    x = np.linspace(x_min, x_max, 2000)
    y = dist.pdf(x)

    # Критические значения
    if tails == 'two':
        cv_right = dist.ppf(1 - alpha/2)
        cv_left  = dist.ppf(alpha/2)
    elif tails == 'right':
        cv_right = dist.ppf(1 - alpha)
        cv_left = None
    elif tails == 'left':
        cv_left = dist.ppf(alpha)
        cv_right = None
    else:
        raise ValueError("tails должен быть 'two', 'left' или 'right'.")

    # p-value
    if tails == 'two':
        p_value = 2 * min(dist.cdf(stat), 1 - dist.cdf(stat))
    elif tails == 'right':
        p_value = 1 - dist.cdf(stat)
    else:  # left
        p_value = dist.cdf(stat)

    reject = p_value < alpha

    # Построение графика
    plt.figure(figsize=(9, 5))
    plt.plot(x, y, color='black', lw=1.5, label='Плотность при H0')

    # Заштриховать критические области (красным)
    if tails == 'two':
        mask_left = x <= cv_left
        mask_right = x >= cv_right
        plt.fill_between(x[mask_left], 0, y[mask_left], color='red', alpha=0.3, label='Критическая область')
        plt.fill_between(x[mask_right], 0, y[mask_right], color='red', alpha=0.3)
        plt.axvline(cv_left, color='red', ls='--')
        plt.axvline(cv_right, color='red', ls='--')
    elif tails == 'right':
        mask_right = x >= cv_right
        plt.fill_between(x[mask_right], 0, y[mask_right], color='red', alpha=0.3, label='Критическая область')
        plt.axvline(cv_right, color='red', ls='--')
    else:  # left
        mask_left = x <= cv_left
        plt.fill_between(x[mask_left], 0, y[mask_left], color='red', alpha='0.3', label='Критическая область')
        plt.axvline(cv_left, color='red', ls='--')

    # Линия наблюдаемой статистики
    plt.axvline(stat, color='blue', lw=2, label=f'Статистика = {stat:.3f}')

    # Заштриховать p-value область (синей штриховкой)
    if tails == 'two':
        if stat >= 0:
            mask_p1 = x >= stat
            sym = -abs(stat)
            mask_p2 = x <= sym
        else:
            mask_p1 = x <= stat
            sym = abs(stat)
            mask_p2 = x >= sym
        plt.fill_between(x[mask_p1], 0, y[mask_p1], color='blue', alpha=0.25, hatch='//', edgecolor='blue', linewidth=0)
        plt.fill_between(x[mask_p2], 0, y[mask_p2], color='blue', alpha=0.25, hatch='//', edgecolor='blue', linewidth=0)
    elif tails == 'right':
        mask_p = x >= stat
        plt.fill_between(x[mask_p], 0, y[mask_p], color='blue', alpha=0.25, hatch='//', edgecolor='blue', linewidth=0)
    else:  # left
        mask_p = x <= stat
        plt.fill_between(x[mask_p], 0, y[mask_p], color='blue', alpha=0.25, hatch='//', edgecolor='blue', linewidth=0)

    # Подписи и решение
    title = (f"{'z' if test=='z' else 't'}-тест, {('двусторонний' if tails=='two' else 'правосторонний' if tails=='right' else 'левосторонний')}, "
             f"alpha={alpha:.3f}, p-value={p_value:.4g}. "
             f"{'Отклоняем H0' if reject else 'Не отклоняем H0'}.")
    plt.title(title)
    plt.xlabel('Статистика теста под H0')
    plt.ylabel('Плотность')
    plt.legend(loc='upper left')
    plt.tight_layout()
    plt.show()

# Пример: двусторонний z-тест, alpha=0.05, наблюдаемая статистика z=2.1
plot_hypothesis_test(test='z', tails='two', alpha=0.05, stat=2.1)

# Пример t-теста (раскомментируйте при необходимости)
# plot_hypothesis_test(test='t', tails='right', alpha=0.01, stat=2.6, df=18)
```

```python
from statsmodels.stats.descriptivestats import sign_test

# задаём уровень значимости
alpha = 0.05 

# загружаем данные
data = [55, 53, 60, 49, 45, 57, 46, 53, 59, 53, 53, 55, 42, 41, 59, 43, 47, 60, 50, 57, 59, 56, 52, 48, 59, 53, 59, 50, 59, 59]

# проводим тест
_, p = sign_test(data, 60)

print('p-value = {:.3f}'.format(p))

# интерпретируем результат
if p <= alpha:
    print('p-значение меньше, чем заданный уровень значимости {:.2f}. Отвергаем нулевую гипотезу.'.format(alpha))
else:
    print('p-значение больше, чем заданный уровень значимости {:.2f}. У нас нет оснований отвергнуть нулевую гипотезу.'.format(alpha))
```

```python
from statsmodels.stats.contingency_tables import mcnemar
import pandas as pd

# задаём уровень значимости
alpha = 0.05 

data = pd.read_csv(r'C:\Users\user\Desktop\Skilfactory\data\cat_variables_usage.csv')

# вычисляем таблицу сопряжённости
table = pd.crosstab(data['До'], data['После'])

# проводим тест
res = mcnemar(table)

# извлекаем значение p-value из результатов теста
p = res.pvalue

print('p-value = {:.3f}'.format(p))

# интерпретируем результат
if p <= alpha:
    print('p-значение меньше, чем заданный уровень значимости {:.2f}. Отвергаем нулевую гипотезу.'.format(alpha))
else:
    print('p-значение больше, чем заданный уровень значимости {:.2f}. У нас нет оснований отвергнуть нулевую гипотезу.'.format(alpha))
```
